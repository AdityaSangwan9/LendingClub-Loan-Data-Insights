```python
# Loan Default Risk Analysis - Lending Club Dataset
# Author: Aditya

# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load dataset (sample CSV from Kaggle or a subset you saved)
df = pd.read_csv("lending_club_sample.csv")

# 3. Inspect data
print(df.head())
print(df.info())

# 4. Data Cleaning
# Remove rows with missing loan status or amount
df = df.dropna(subset=["loan_status", "loan_amnt"])

# Convert interest rate to numeric
df['int_rate'] = df['int_rate'].str.replace('%','').astype(float)

# Standardize loan purpose labels
df['purpose'] = df['purpose'].str.lower().str.strip()

# Create default flag
df['default_flag'] = df['loan_status'].apply(lambda x: 1 if x in ["Charged Off", "Default"] else 0)

# Group income into ranges
bins = [0, 25000, 50000, 75000, 100000, 150000]
labels = ['0-25k','25-50k','50-75k','75-100k','100k+']
df['income_group'] = pd.cut(df['annual_inc'], bins=bins, labels=labels)

# 5. Analysis & Insights
# Default rate by credit grade
grade_default = df.groupby('grade')['default_flag'].mean()
print(grade_default)

# Default rate by loan purpose
purpose_default = df.groupby('purpose')['default_flag'].mean()
print(purpose_default)

# Default rate by loan term
term_default = df.groupby('term')['default_flag'].mean()
print(term_default)

# Default rate by income group
income_default = df.groupby('income_group')['default_flag'].mean()
print(income_default)

# 6. Visualization
plt.figure(figsize=(8,5))
sns.barplot(x=grade_default.index, y=grade_default.values)
plt.title("Default Rate by Credit Grade")
plt.ylabel("Default Rate")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x=purpose_default.index, y=purpose_default.values)
plt.title("Default Rate by Loan Purpose")
plt.xticks(rotation=45)
plt.show()

# 7. Conclusion
# Summarize findings in markdown cells:
# - Lower credit grades default more
# - Small business loans riskier
# - Longer loan terms default more
# - Higher interest rates align with higher defaults
# - Lowest income group defaults more
```
